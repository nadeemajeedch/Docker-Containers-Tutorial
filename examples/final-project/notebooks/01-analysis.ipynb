# Final project analysis

Load the CSV, fit a linear model, save a figure under `/workspace/results`.

```python
import os
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression

print("student", os.environ.get("STUDENT_NAME"))
table = pd.read_csv("/workspace/data/study_hours.csv")
model = LinearRegression().fit(table[["hours"]], table["score"])
print("predicted at 5 hours", float(model.predict([[5.0]])[0]))

Path("/workspace/results").mkdir(parents=True, exist_ok=True)
fig, ax = plt.subplots()
ax.scatter(table["hours"], table["score"])
ax.plot(table["hours"], model.predict(table[["hours"]]), color="tab:orange")
ax.set_title("Final project: hours vs score")
fig.savefig("/workspace/results/notebook-plot.png", dpi=120)
plt.show()
```
