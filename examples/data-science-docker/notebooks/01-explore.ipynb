# Explore study hours

This notebook reads `/workspace/data/study_hours.csv` from the bind mount.

```python
import os
from pathlib import Path

import pandas as pd

print("STUDENT_NAME", os.environ.get("STUDENT_NAME"))
table = pd.read_csv("/workspace/data/study_hours.csv")
table
```

```python
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

model = LinearRegression().fit(table[["hours"]], table["score"])
print("predicted at 5 hours", float(model.predict([[5.0]])[0]))

fig, ax = plt.subplots()
ax.scatter(table["hours"], table["score"])
ax.plot(table["hours"], model.predict(table[["hours"]]), color="tab:orange")
ax.set_title("Study hours vs score")
plt.show()

Path("/workspace/results").mkdir(parents=True, exist_ok=True)
```
